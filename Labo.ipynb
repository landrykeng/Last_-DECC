```python
from pathlib import Path

def afficher_arborescence(chemin=Path.cwd(), prefixe=""):
    chemin = Path(chemin)
    print(f"{prefixe}{chemin.name}/")
    elements = sorted(chemin.iterdir(), key=lambda p: (p.is_file(), p.name.lower()))

    for i, element in enumerate(elements):
        dernier = i == len(elements) - 1
        branche = "└── " if dernier else "├── "
        print(f"{prefixe}{branche}{element.name}{'/' if element.is_dir() else ''}")
        #if element.is_dir():
        #    afficher_arborescence(element, prefixe + ("    " if dernier else "│   "))

afficher_arborescence()
```

```python
import pandas as pd
from pathlib import Path

RACINE = Path(r"C:\Users\LK\Desktop\DECC")

dataframes = {}

for dossier in RACINE.iterdir():
    
    print(dossier.name)

    if not dossier.is_dir():
        continue

    fichiers = list(dossier.rglob("*.xlsx"))

    tables = []

    for fichier in fichiers:
        try:
            df = pd.read_excel(fichier)
            tables.append(df)
            print(f"    {fichier} → {len(df)} lignes")
        except Exception as e:
            print(f"Erreur : {fichier} → {e}")

    if tables:

        df_final = pd.concat(
            tables,
            ignore_index=True,
            sort=False
        )

        dataframes[dossier.name] = df_final

        print(
            f"{dossier.name} : "
            f"{len(fichiers)} fichiers, "
            f"{len(df_final)} lignes"
        )
```

```python
#df_bepc=dataframes["BEPC"]
df_cap_indus=dataframes["CAP INDUSTRIEL"]
df_cap_stt=dataframes["CAP STT"]
df_indstrial_cap=dataframes["INDUSTRIAL CAP"]
df_tst_cap=dataframes["TST CAP"]
```

```python
df_cap_indus.to_csv(r"C:\Users\LK\Desktop\DECC\CAP_INDUSTRIEL.csv", index=False)
df_cap_stt.to_csv(r"C:\Users\LK\Desktop\DECC\CAP_STT.csv", index=False)
df_indstrial_cap.to_csv(r"C:\Users\LK\Desktop\DECC\INDUSTRIAL_CAP.csv", index=False)
df_tst_cap.to_csv(r"C:\Users\LK\Desktop\DECC\TST_CAP.csv", index=False)
```
